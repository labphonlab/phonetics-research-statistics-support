■ なぜColab環境から始めるのか：

Rを何もインストールせずに使えること以外にも、Colabを選ぶ理由は4つある。

1. **環境構築で挫折しない**——RやRStudioの自前インストールはOSやバージョンの組み合わせでつまずきやすい。統計を学ぶ前に環境構築で心が折れる読者をできるだけ出したくない
2. **どの端末でも同じ結果になる**——WindowsでもMacでも、ブラウザさえあれば同じノートブックが同じように動く
3. **無料で計算機を使える**——第8章のベイズモデルのような重い処理もGoogle側の計算資源で動く（長時間・大規模な計算には利用制限があるが、本書の演習規模では問題にならない）
4. **配布・共有が容易**——本書の各章はURL一つで届き、開けば自分のGoogleアカウント上にコピーができてすぐ書き換えられる

**ただしColabは「学習を始めるための環境」であって「一生そこに留まるべき環境」ではない。** 自分の研究データを本格的に扱うようになったら、ローカル環境（自分のパソコン上のR）に移る場面が出てくる——大きなコーパスを扱うとColabのディスク容量・実行時間の制限に当たりやすいこと、Gitでのコード管理はローカルのほうが自然に組み込めること、Colabの無料枠には計算資源の上限があることが理由である。

**移植手順（本書に慣れたあとで）**
1. [R本体](https://cran.r-project.org/)→[RStudio Desktop](https://posit.co/download/rstudio-desktop/)の順にインストール（無料版で十分）
2. 各章冒頭の`install.packages(...)`をRStudioのコンソールで1回実行する（Colabと違い、次回以降は`library(...)`だけで済む）
3. データの読み込みを、`set.seed(42)`でその場生成する代わりに`read_csv("~/Documents/measurements.csv")`のようなローカルパス指定に変える（Google Driveのマウント操作は不要になる）
4. RStudioの「New Project」で研究ごとにフォルダを分けて管理する
5. このノートブックのセルの中身（`#`コメントも含めて）は文法がそのままのRなので、`.R`や`.Rmd`にコピペするだけで動く

つまり「Colabで基礎を身につけ、ローカルに移植して本格的な研究へ」という順番を想定している。

■ プログラム・スクリプト・ノートブックとは：

**プログラム（スクリプト）**とは、コンピュータへの指示を順番に並べた文章のことである。人間が読む文章と同じように上から下へ実行されるが、コンピュータが理解できる決まった書き方（文法）に従う必要がある。

**Colabノートブック**は、このプログラムを「セル」という小さな塊に分けて、1つずつ実行できるようにしたものである。セルには2種類ある。
- **コードセル**（このノートブックで灰色の背景の部分）: 実際にRのプログラムが書いてあり、再生ボタン（▷）を押すかShift+Enterで実行できる
- **markdownセル**（このセルのように文章だけの部分）: 説明文。実行しても何も起きない

**上から順番に実行すること。** 後の回のコードは前の回で作った変数（例えば`vot_data`というデータ）を前提にしていることが多いので、セルを飛ばしたり順番を入れ替えたりすると「そんな変数はない」というエラーになる。

■ 変数（データに名前をつける）：

Rでは `<-` という記号（マイナスと不等号を組み合わせたもの、「矢印」と読む）を使って、右側の値に左側の名前をつける。

```r
x <- 5       # xという名前に5という数値を割り当てる（「代入」と呼ぶ）
y <- 10
x + y        # xとyを足す。この行だけがセルの最後にあると、その結果が画面に表示される

# --- 実行結果 ---
# [1] 15
#
# 【出力の読み方】
# ・行頭の [1] は値ではなく「この行の最初の要素が全体の何番目か」を示す番号である。
#   Rは結果をベクトル（値の並び）として返すので、1個しかない場合も [1] が付く。
#   したがって表示されている中身は 15 だけ。x + y = 5 + 10 の答えである。
# ・x <- 5 と y <- 10 の行には何も表示されない。代入は「値をしまう」操作であって、
#   結果を返す操作ではないため。セルの中で表示されるのは、代入せずに
#   書いた行（ここでは x + y）の結果だけである。
```

■ ベクトルとデータフレーム（複数の値をまとめて扱う）：

`c()`は複数の値を1本にまとめる関数（「combine」の頭文字）。音声学のデータは通常、複数の観測値（複数の発話・複数の話者）を同時に扱うので、この「まとめる」操作が基本になる。

```r
votvalues <- c(15, 78, 22, 65)   # 4つの数値を1本のベクトルにまとめる
votvalues              # 中身を確認
mean(votvalues)        # mean()は平均を計算する関数

# --- 実行結果 ---
# [1] 15 78 22 65
# [1] 45
#
# 【出力の読み方】
# ・1行目 [1] 15 78 22 65 は votvalues の中身そのもの。4つの値が1本の
#   ベクトルとして横に並んでいる。[1] は「この行の先頭が1番目の要素」の意味で、
#   要素数が多くて複数行に折り返されると、2行目の行頭は [5] のようになる。
# ・2行目 [1] 45 が mean(votvalues) の結果。(15+78+22+65)/4 = 45 である。
#   平均は1個の数値なので、やはり [1] が付いた1行で返る。

# 表形式（行=観測、列=変数）にしたものが「データフレーム」または「tibble」。
# 第1章以降で使う vot_data はまさにこの形をしている。
mini_table <- data.frame(
  word = c("pa", "ba", "ta", "da"),
  VOT = votvalues
)
mini_table

# --- 実行結果 ---
#   word VOT
# 1   pa  15
# 2   ba  78
# 3   ta  22
# 4   da  65
#
# 【出力の読み方】
# ・1行目の word / VOT は列名（変数名）。データフレームは「列＝変数、行＝観測」
#   という形をしており、この表なら「4つの語について、それぞれVOTを1回測った」
#   と読む。本書で以降ずっと使う vot_data も、行数と列数が違うだけで同じ形。
# ・左端の 1 2 3 4 は行番号（行の名前）であって、データの中身ではない。
#   列として数えないこと。
# ・word列の値にクォートが付いていないが、中身は文字列である。型を確かめたい
#   ときは str(mini_table) や glimpse(mini_table) を使う。
```

■ 関数（決まった処理をひとまとめにしたもの）：

`mean(votvalues)`のように、「関数名(渡す値)」という書き方をする。関数は「入力を受け取って、決まった処理をして、結果を返す機械」だと考えるとよい。`mean()`はレシピでいう「材料を全部合計して人数で割る」という手順を、毎回自分で書かなくても呼び出せるようにしたものである。本書で使うほぼすべてのコードは、こうした関数の呼び出しの組み合わせでできている（例: `lm()`＝回帰分析をする関数、`ggplot()`＝グラフを描く関数）。

■ パッケージ（追加機能を持ってくる）：

Rの基本機能だけでは、統計解析やグラフ作成に便利な関数が足りないことが多い。**パッケージ**とは、誰かが作った関数の詰め合わせで、スマートフォンのアプリのようなものだと考えるとよい。

- `install.packages("tidyverse")`: パッケージを自分の環境にダウンロードする（アプリのインストールに相当。1回だけでよい）
- `library(tidyverse)`: すでにインストール済みのパッケージを「今この場で使えるようにする」（アプリを開くことに相当。Colabは実行環境が毎回リセットされるため、本書では毎回両方書いてある）

■ パイプ演算子 %>%（処理をつなげる）：

第2章以降、`データ %>% 関数A() %>% 関数B()`という書き方が繰り返し出てくる。`%>%`は「左側の結果を、右側の関数の最初の引数として渡す」という意味で、**「それから」**と読み替えると理解しやすい。

```r
suppressMessages(library(tidyverse))  # パイプ演算子はtidyverseに含まれる

c(15, 78, 22, 65) %>%   # このベクトルを、それから
  mean() %>%             # 平均を取り、それから
  round(1)                # 小数第1位に丸める

# --- 実行結果 ---
# [1] 45
#
# 【出力の読み方】
# ・パイプ %>% は「左の結果を、右の関数の第1引数として渡す」だけの記号なので、
#   出力は素朴に書いた場合と完全に同じ [1] 45 になる。
#   c(15,78,22,65) → mean() で 45 → round(45, 1) で 45（小数第1位まで丸めても
#   45のままなので 45.0 とは表示されない）。
# ・このセルは同じ値を2回計算しているので、実行すると [1] 45 が2回表示される。
#   1回目がパイプ版、2回目が round(mean(...), 1) 版の結果である。
#   **2つが一致していることが、パイプが特別な計算をしていない証拠**になる。

# 上のコードは次と全く同じ意味だが、処理が増えるほど%>%の方が読みやすくなる：
round(mean(c(15, 78, 22, 65)), 1)
```

■ よくあるエラーの読み方：

初めてエラーメッセージを見ると身構えてしまうが、Rのエラーはだいたい次のどれかである。

- `object 'vot_data' not found` → 前のセルを実行し忘れている（変数がまだ作られていない）
- `could not find function "xxx"` → そのパッケージを`library()`し忘れている
- `unexpected ')' `や`unexpected symbol` → 括弧の対応や、カンマの打ち忘れなど、書き方の間違い（構文エラー）
- `Error in ... : non-numeric argument` → 本来数値であるべき列が文字列（`<chr>`）のままになっている

エラーが出たら、まず「どのセルの、何行目か」を確認し、次に「そのすぐ上のセルまで、順番どおりに全部実行したか」を確認する——これだけで大半のエラーは解決する。

■ 演習：

1. `x <- 3` として、`x`を2乗した値（`x^2`）を計算せよ。
2. `c(10, 20, 30, 40, 50)`というベクトルを作り、`sum()`（合計）と`length()`（要素数）をそれぞれ試せ。
3. わざと閉じ括弧を1つ消してセルを実行し、どんなエラーメッセージが出るか確認せよ（その後、括弧を戻して直せることを確認する）。

■ 演習の解答：

以下のセルを実行すると、上の3問の答えがそのまま確認できる。**まず自分で書いてみてから**開くこと（答えを読むだけでは書けるようにならない）。

```r
# 1. x <- 3 として x^2 を計算する
x <- 3
x^2        # ^ はべき乗の記号。x^2 は x を2回かけた値

# 2. ベクトルに sum() と length() を試す
v <- c(10, 20, 30, 40, 50)
sum(v)     # 合計
length(v)  # 要素の個数
mean(v)    # 参考: sum(v) / length(v) と同じ値になる

# 3. わざと閉じ括弧を1つ消すとどうなるかを、セルを止めずに確かめる。
#    try() で囲むと、エラーが起きてもそこで実行が止まらず次の行に進む。
#    （本当にセルに書いて実行すると、そのセルはそこで停止する）
try(parse(text = "mean(c(10, 20, 30)"))

# --- 実行結果 ---
# [1] 9
# [1] 150
# [1] 5
# [1] 30
# Error in parse(text = "mean(c(10, 20, 30)") :
#   <text>:2:0: unexpected end of input
# 1: mean(c(10, 20, 30)
#    ^
#
# 【出力の読み方】
# ・[1] 9   … 1問目。x <- 3 なので x^2 = 9。
# ・[1] 150 … 2問目の sum(v)。10+20+30+40+50 = 150。
# ・[1] 5   … 同じく length(v)。要素は5個。
# ・[1] 30  … 参考で入れた mean(v)。150 / 5 = 30 と一致する。
#   **合計・個数・平均の3つが辻褄の合う関係になっている**ことを確かめておくと、
#   関数が自分の思ったとおりに動いているかの簡単な検算になる。
# ・最後のエラーが3問目。"unexpected end of input"（入力が途中で終わっている）は
#   **括弧が閉じていない**ときの典型的なメッセージである。続く2行がさらに親切で、
#   問題のある行の中身と、どこで行き詰まったかを ^ で指している。
#   エラーが出たら、まずこの ^ の位置と直前の括弧の対応を数えること。
#   なお、Rの表示言語が日本語に設定されている環境では
#   「想定外の入力の終端です」と訳されて出る。文言は違っても意味は同じ。
```
